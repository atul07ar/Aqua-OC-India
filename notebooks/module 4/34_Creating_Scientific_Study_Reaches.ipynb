```python
# ==========================================
# Lesson 34
# Scientific Study Reaches
# ==========================================

import ee
import geemap
import pandas as pd

ee.Initialize(project="oc-flux")

Map = geemap.Map()

Map
```

```python
Create a Database

Run:
```

```python
ganga = pd.DataFrame({

"Reach":[

"Rishikesh",
"Haridwar",
"Kanpur",
"Prayagraj",
"Varanasi",
"Patna",
"Bhagalpur",
"Farakka"

],

"Lon":[

78.2676,
78.1642,
80.3319,
81.8463,
83.0120,
85.1376,
87.0025,
88.0300

],

"Lat":[

30.0869,
29.9457,
26.4499,
25.4358,
25.3176,
25.5941,
25.2425,
24.8100

]

})

ganga
```

```python
Display the Reaches
```

```python
Map = geemap.Map()

for _, row in ganga.iterrows():

    pt = ee.Geometry.Point(
        [row["Lon"], row["Lat"]]
    )

    Map.addLayer(
        pt,
        {"color":"red"},
        row["Reach"]
    )

Map.centerObject(
    ee.Geometry.Point([82,26]),
    6
)

Map
```

```python
Create Buffers

Each point becomes a study reach.
Each yellow circle represents a 5 km buffer around a representative reach.
```

```python
Map = geemap.Map()

for _, row in ganga.iterrows():

    point = ee.Geometry.Point(
        [row["Lon"], row["Lat"]]
    )

    reach = point.buffer(5000)

    Map.addLayer(
        reach,
        {"color":"yellow"},
        row["Reach"]
    )

Map.centerObject(
    ee.Geometry.Point([82,26]),
    6
)

Map
```

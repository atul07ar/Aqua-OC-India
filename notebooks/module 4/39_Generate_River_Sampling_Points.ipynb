```python
import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()
```

```python
Load HydroRIVERS
```

```python
rivers = ee.FeatureCollection(
    "projects/oc-flux/assets/HydroRIVERS_v10_asi_shp"
)
```

```python
Define ROI
```

```python
roi = ee.Geometry.Point(
    [79.8033,11.1313]
).buffer(10000)
```

```python
Filter River
```

```python
river = rivers.filterBounds(roi)
```

```python
Select Main Channel
```

```python
mainRiver = ee.Feature(
    river.sort("DIS_AV_CMS", False).first()
)
```

```python
Display
```

```python
Map.centerObject(mainRiver,11)

Map.addLayer(
    mainRiver,
    {"color":"blue"},
    "Main River"
)

Map
```

```python
Get Geometry
```

```python
line = mainRiver.geometry()
```

```python
Measure River Length
```

```python
length = line.length()

print(
    "Length (m):",
    length.getInfo()
)
```

```python
Choose Sampling Interval
```

```python
spacing = 300
```

```python
Calculate Number of Points
```

```python
number = ee.Number(length.divide(spacing)).toInt()

print(number.getInfo())
```

```python
Create Distances
```

```python
distances = ee.List.sequence(
    0,
    length,
    spacing
)
```

```python
Generate Points
```

```python
points = ee.FeatureCollection(

    distances.map(

        lambda d:

        ee.Feature(

            line.interpolate(
                ee.Number(d)
            )

        )

    )

)
```

```python
Display Points
```

```python
Map.addLayer(
    points,
    {"color":"red"},
    "Sampling Points"
)

Map
```

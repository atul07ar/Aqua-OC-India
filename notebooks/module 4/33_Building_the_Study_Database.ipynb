```python
Add Approximate Coordinates

For now we'll use one representative location for each river. Later, we'll replace these with official river reaches.
```

```python
rivers = pd.DataFrame({

"River":[

"Ganga",
"Brahmaputra",
"Godavari",
"Krishna",
"Mahanadi",
"Narmada",
"cauvery"

],

"Lon":[

83.0120,
91.7440,
81.7830,
80.6480,
85.8500,
75.7000,
79.1400

],

"Lat":[

25.3176,
26.2000,
17.0000,
16.6000,
20.3000,
22.3000,
11.8700

]

})

rivers
```

```python
Display Rivers on the Map
```

```python
Map = geemap.Map()

for _, row in rivers.iterrows():

    point = ee.Geometry.Point([
        row["Lon"],
        row["Lat"]
    ])

    Map.addLayer(
        point,
        {"color":"red"},
        row["River"]
    )

Map.centerObject(
    ee.Geometry.Point([82,22]),
    4
)

Map
```

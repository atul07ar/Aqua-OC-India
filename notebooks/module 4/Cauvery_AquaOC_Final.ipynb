````python
# Section 8
## Cloud Masking and Surface Reflectance Scaling

### Purpose

Raw Landsat imagery contains clouds, cloud shadows, cirrus clouds, and sensor scaling values that must be corrected before scientific analysis.

In this section, the Landsat Quality Assessment (QA_PIXEL) band is used to identify and remove:

- Clouds
- Cloud Shadows
- Cirrus Clouds

After cloud masking, the official USGS scaling factors are applied to convert the digital numbers into **Surface Reflectance** values.

Finally, only the spectral bands required for Aqua-OC modelling are retained.

This preprocessing step ensures that all extracted reflectance values represent true surface conditions and are suitable for water quality analysis.
````

````python
# ============================================================
# Section 8
# Cloud Masking and Surface Reflectance Scaling
# ============================================================

def mask_landsat(image):
    """
    Cloud masking and scaling for Landsat Collection 2 Level-2
    """

    # QA_PIXEL band
    qa = image.select("QA_PIXEL")

    # Bit masks
    cloud_shadow = 1 << 4
    clouds = 1 << 3
    cirrus = 1 << 2

    # Keep only clear pixels
    mask = (
        qa.bitwiseAnd(cloud_shadow).eq(0)
        .And(qa.bitwiseAnd(clouds).eq(0))
        .And(qa.bitwiseAnd(cirrus).eq(0))
    )

    # Apply official scale factor
    optical = (
        image.select("SR_B.*")
        .multiply(0.0000275)
        .add(-0.2)
    )

    return (
        image.addBands(optical, overwrite=True)
             .updateMask(mask)
    )
````

````python
# ============================================================
# Apply Cloud Masking
# ============================================================

landsat_masked = landsat.map(mask_landsat)

print("Cloud masking applied successfully.")
````

````python
# Section 9
## Create Annual Cloud-Free Surface Reflectance Composite

### Purpose

After cloud masking and surface reflectance correction, multiple Landsat images are available for the study area throughout the year. Since individual satellite scenes may still contain residual clouds or temporal variability, a single representative image is required for analysis.

In this section, all cloud-free Landsat 8 and Landsat 9 images acquired during the study period are combined into a single annual composite using the **median reducer**.

The median composite:

- Minimizes the influence of remaining clouds and atmospheric noise.
- Produces a stable representation of surface reflectance.
- Provides consistent spectral values for extracting reflectance at the 50 km sampling points.
- Serves as the primary input for Aqua-OC POC estimation.

The composite is clipped to the 5 km Cauvery River buffer to reduce computational requirements and focus only on the study area.
````

````python
# ============================================================
# Section 9
# Create Annual Median Composite
# ============================================================

# Create median composite
composite = (
    landsat_masked
    .median()
    .clip(buffer_ee)
)

print("Annual median composite created successfully.")
````

````python
# Section 11
## Extract Landsat Surface Reflectance at 50 km Sampling Points

### Purpose

The objective of this section is to extract the annual Landsat Surface Reflectance values from the cloud-free composite at each 50 km sampling point along the Cauvery River.

The extracted reflectance values represent the spectral characteristics of the river surface and serve as predictor variables for the Aqua-OC Random Forest model.

For every sampling point, the following bands are extracted:

- SR_B1 (Coastal Aerosol)
- SR_B2 (Blue)
- SR_B3 (Green)
- SR_B4 (Red)
- SR_B5 (Near Infrared)
- SR_B6 (Shortwave Infrared 1)
- SR_B7 (Shortwave Infrared 2)

The extracted dataset will be exported for subsequent POC estimation.
````

````python
# ============================================================
# Section 11
# Extract Surface Reflectance at Sampling Points
# ============================================================

reflectance = reflectance_image.sampleRegions(
    collection=points_ee,
    scale=30,
    geometries=True
)

print("Surface Reflectance extracted successfully.")
````

````python
# ============================================================
# Number of Extracted Features
# ============================================================

print("Number of observations:")

print(reflectance.size().getInfo())
````

````python
# ============================================================
# Convert Earth Engine FeatureCollection to Pandas DataFrame
# Compatible with pandas 3.x
# ============================================================

import pandas as pd

# Get FeatureCollection as a dictionary
features = reflectance.getInfo()["features"]

records = []

for feature in features:

    props = feature["properties"].copy()

    # Add coordinates
    lon, lat = feature["geometry"]["coordinates"]

    props["Longitude"] = lon
    props["Latitude"] = lat

    records.append(props)

# Create DataFrame
df = pd.DataFrame(records)

print(df.head())

print()
print("Number of observations:", len(df))
````

```python
import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()

rivers = ee.FeatureCollection(
    "projects/oc-flux/assets/HydroRIVERS_v10_asi_shp"
)

# Reference point on the Cauvery
pt = ee.Geometry.Point([76.95, 11.55])

# Only rivers within 5 km
small = rivers.filterBounds(
    pt.buffer(5000)
)

Map.centerObject(pt, 10)

Map.addLayer(
    small,
    {"color":"blue"},
    "Nearby Rivers"
)

Map
```

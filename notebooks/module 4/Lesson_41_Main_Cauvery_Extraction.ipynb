```python
import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()

# Load HydroRIVERS
rivers = ee.FeatureCollection(
    "projects/oc-flux/assets/HydroRIVERS_v10_asi_shp"
)
```

```python
Extract only the Main Cauvery
```

```python
cauvery = rivers.filter(
    ee.Filter.eq("MAIN_RIV", 41376077)
)

print("Segments:", cauvery.size().getInfo())
```

```python
Display the river
```

```python
Map.centerObject(cauvery, 8)

Map.addLayer(
    cauvery,
    {"color": "blue"},
    "Main Cauvery"
)

Map
```

```python
Merge all segments into one river
```

```python
river = cauvery.geometry()

Map.addLayer(
    river,
    {"color": "red"},
    "Merged River"
)

Map
```

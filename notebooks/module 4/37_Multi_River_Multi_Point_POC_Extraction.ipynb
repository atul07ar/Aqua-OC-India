```python
import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()
```

```python
Create Sampling Points

Let's start with the Ganga.
```

```python
ganga = ee.FeatureCollection([

ee.Feature(
    ee.Geometry.Point([79.00,30.00]),
    {"River":"Ganga","Point":"G1"}
),

ee.Feature(
    ee.Geometry.Point([79.55,29.95]),
    {"River":"Ganga","Point":"G2"}
),

ee.Feature(
    ee.Geometry.Point([80.35,28.90]),
    {"River":"Ganga","Point":"G3"}
),

ee.Feature(
    ee.Geometry.Point([81.90,27.90]),
    {"River":"Ganga","Point":"G4"}
),

ee.Feature(
    ee.Geometry.Point([83.00,26.80]),
    {"River":"Ganga","Point":"G5"}
),

ee.Feature(
    ee.Geometry.Point([85.15,25.60]),
    {"River":"Ganga","Point":"G6"}
),

ee.Feature(
    ee.Geometry.Point([87.80,24.90]),
    {"River":"Ganga","Point":"G7"}
),

ee.Feature(
    ee.Geometry.Point([88.35,22.60]),
    {"River":"Ganga","Point":"G8"}
)

])
```

```python
Display the Points
```

```python
Map.centerObject(ganga,6)

Map.addLayer(
    ganga,
    {"color":"red"},
    "Ganga Points"
)

Map
```

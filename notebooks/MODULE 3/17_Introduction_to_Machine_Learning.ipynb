```python
from sklearn.ensemble import RandomForestRegressor
import numpy as np

# Example input (Red Band reflectance)
X = np.array([[0.10],
              [0.20],
              [0.30],
              [0.40]])

# Example measured POC
y = np.array([2.5, 4.0, 6.2, 8.0])

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

prediction = model.predict([[0.25]])

print("Predicted POC:", prediction)
```

```python
You should see a predicted value close to the training pattern.
The exact number may vary slightly because Random Forest builds many decision trees.
```

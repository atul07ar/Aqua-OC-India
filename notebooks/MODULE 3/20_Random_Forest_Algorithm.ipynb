```python
import numpy as np
from sklearn.ensemble import RandomForestRegressor

# Feature (Red Band)
X = np.array([
    [0.10],
    [0.15],
    [0.20],
    [0.25],
    [0.30]
])

# Target (POC)
y = np.array([
    2,
    3,
    5,
    7,
    9
])

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

prediction = model.predict([[0.23]])

print(prediction)
```

```python
Python Demonstration
Observe how predictions change as the Red band value changes.
```

```python
X = np.array([
    [0.10],
    [0.15],
    [0.20],
    [0.25],
    [0.30]
])

y = np.array([2,3,5,7,9])

model = RandomForestRegressor(
    n_estimators=500,
    random_state=42
)

model.fit(X,y)

for value in [0.12,0.18,0.24,0.29]:
    pred = model.predict([[value]])
    print(f"Red Band = {value:.2f} → Predicted POC = {pred[0]:.2f}")
```

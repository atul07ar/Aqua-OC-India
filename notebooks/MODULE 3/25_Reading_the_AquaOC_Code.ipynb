```python
Load Satellite Images
Think of an ImageCollection as a folder containing hundreds of Landsat images.
```

```python
collection = (
    ee.ImageCollection(
        "LANDSAT/LC08/C02/T1_L2"
    )
)
```

```python
Filter Images
The code reduces the collection to:

one location
one year

instead of thousands of images from around the world.
```

```python
collection = (
    collection
    .filterBounds(point)
    .filterDate(
        "2023-01-01",
        "2023-12-31"
    )
)
```

```python
Create Composite
Instead of processing every image individually,

we combine them into one representative image.
```

```python
image = collection.median()
```

```python
collection = collection.map(
    mask_cloud
)
```

```python
ndwi = image.normalizedDifference(
    ["SR_B3","SR_B5"]
)
```

```python
Understanding Functions

Example
The function accepts an image,

performs one task,

and returns the result.
```

```python
def calculate_ndwi(image):

    ndwi = image.normalizedDifference(
        ["SR_B3","SR_B5"]
    )

    return ndwi
```

```python
Later
```

```python
Real Aqua-OC Logic

A simplified version looks like this.
Load Image

↓

Cloud Mask

↓

Water Mask

↓

Calculate Indices

↓

Feature Stack

↓

Random Forest

↓

Prediction

↓

Display

Notice that every line of code belongs to one of these steps.
```

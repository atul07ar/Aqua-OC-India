# Section 3 : MOREPOC Database Description

## Overview

The **MOREPOC (Multi-Source River Particulate Organic Carbon)** database is a global compilation of field measurements of riverine Particulate Organic Carbon (POC) collected from rivers across different climatic, hydrological, and geographical regions.

The dataset integrates measurements reported in published scientific literature and provides standardized information on river water quality, organic carbon concentration, and sampling locations.

The MOREPOC database serves as the reference (ground truth) dataset for developing remote sensing and machine learning models capable of estimating riverine POC from satellite observations.

---

## Dataset Contents

Each observation in the MOREPOC database represents one field sampling event and typically contains:

- River name
- Basin name
- Country
- Latitude
- Longitude
- Sampling date
- Measured POC concentration
- Suspended sediment concentration
- Additional geochemical variables
- Reference publication

---

## Spatial Coverage

The database contains observations from rivers distributed across multiple continents including:

- Asia
- Europe
- North America
- South America
- Africa
- Australia

This broad geographical coverage makes MOREPOC suitable for developing globally applicable machine learning models.

---

## Temporal Coverage

The observations span several decades, allowing analysis across different hydrological conditions and climatic periods.

Because Landsat imagery has been available since the 1980s, a large proportion of these observations can potentially be matched with satellite data.

---

## Role in This Research

In this dissertation, the MOREPOC database provides:

- Ground observations of riverine POC.
- Geographic locations for satellite image extraction.
- Target values for machine learning.
- Validation data for model performance assessment.

The measured POC concentration serves as the dependent variable, while Landsat-derived spectral information forms the predictor variables.

---

## Advantages of the MOREPOC Database

- Global spatial coverage.
- Literature-validated observations.
- Standardized measurement format.
- Suitable for remote sensing studies.
- Supports machine learning applications.
- Enables comparison across different river systems.

---

## Limitations

Despite its usefulness, the database has several limitations:

- Uneven spatial distribution of samples.
- Missing observations in some regions.
- Temporal mismatch between field sampling and satellite overpass.
- Variable sampling density among rivers.
- Differences in sampling methodologies across published studies.

These limitations are addressed later through quality control, temporal matching, and data preprocessing.

### Interpretation

The MOREPOC database forms the foundation of the Aqua-OC workflow because it provides reliable field measurements of riverine particulate organic carbon collected from diverse river systems worldwide. These observations act as the reference values against which Landsat-derived spectral information will be calibrated using machine learning.

Understanding the characteristics, strengths, and limitations of the dataset is essential before beginning preprocessing and image extraction, as the quality of the final predictive model depends directly on the quality and representativeness of the input data.

Data quality directly influences the accuracy of Landsat image matching and machine learning model performance. Conducting a structured quality assessment ensures that preprocessing decisions are transparent, reproducible, and scientifically justified.

```python
# ============================================================
# Section 9 : Data Quality Assessment
# ============================================================

print("=" * 70)
print("Section 9 : Data Quality Assessment")
print("=" * 70)

# ------------------------------------------------------------
# Dataset Dimensions
# ------------------------------------------------------------

print("\nDataset Shape")
print("-" * 40)
print(morepoc.shape)

# ------------------------------------------------------------
# Missing Values
# ------------------------------------------------------------

missing = morepoc.isnull().sum()

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage (%)": round((missing / len(morepoc)) * 100, 2)
})

print("\nMissing Values")
print("-" * 40)

display(
    missing_df.sort_values(
        "Missing Values",
        ascending=False
    )
)

# ------------------------------------------------------------
# Duplicate Records
# ------------------------------------------------------------

duplicates = morepoc.duplicated().sum()

print("\nDuplicate Records :", duplicates)

# ------------------------------------------------------------
# Duplicate Sample IDs
# ------------------------------------------------------------

if "code" in morepoc.columns:
    duplicate_codes = morepoc["code"].duplicated().sum()
    print("Duplicate Sampling Codes :", duplicate_codes)

# ------------------------------------------------------------
# Coordinate Validation
# ------------------------------------------------------------

invalid_lat = (
    (morepoc["lat"] < -90) |
    (morepoc["lat"] > 90)
).sum()

invalid_lon = (
    (morepoc["lon"] < -180) |
    (morepoc["lon"] > 180)
).sum()

print("\nInvalid Latitude :", invalid_lat)
print("Invalid Longitude:", invalid_lon)

# ------------------------------------------------------------
# Missing Coordinates
# ------------------------------------------------------------

print("\nMissing Latitude :", morepoc["lat"].isna().sum())
print("Missing Longitude:", morepoc["lon"].isna().sum())

# ------------------------------------------------------------
# Target Variable
# ------------------------------------------------------------

print("\nMissing POC Values :", morepoc["conc_poc"].isna().sum())

negative_poc = (
    morepoc["conc_poc"] < 0
).sum()

print("Negative POC Values:", negative_poc)

# ------------------------------------------------------------
# Date Check
# ------------------------------------------------------------

print("\nSample Date Range")
print("-" * 40)

print("Earliest :", morepoc["sample_date"].min())
print("Latest   :", morepoc["sample_date"].max())

print("\nSection 9 Completed Successfully")
```

### Interpretation

The quality assessment indicates that the Indian MOREPOC dataset is generally suitable for remote sensing analysis but requires careful preprocessing before Landsat image extraction and machine learning.

#### Duplicate Records

No completely duplicated records were identified, indicating that each observation represents a unique sampling event.

However, **9 duplicate sampling station codes** were detected. This is expected because the same sampling station may have been visited on different dates or under different hydrological conditions. These records should not be removed unless both the sampling date and measurements are identical.

#### Geographic Coordinates

No invalid latitude or longitude values were found, confirming that all available coordinates lie within valid geographic ranges.

However, **25 observations lack geographic coordinates**. Since satellite image extraction requires precise sampling locations, these observations cannot be linked with Landsat imagery unless their coordinates can be recovered from the original publications or supplementary datasets.

#### Particulate Organic Carbon (POC)

The target variable (`conc_poc`) contains **14 missing observations**. These records cannot be used for supervised machine learning because the response variable is unavailable.

Importantly, **no negative POC concentrations** were identified, indicating that all recorded POC values are physically meaningful.

#### Temporal Coverage

Sampling dates range from **15 March 1997** to **30 September 2013**, providing approximately sixteen years of field observations.

This temporal coverage overlaps well with the Landsat archive, making the dataset suitable for satellite-based analysis.

#### Overall Assessment

The dataset contains reliable observations but requires preprocessing before further analysis. The primary issues identified are:

- Missing geographic coordinates (25 samples)
- Missing POC concentrations (14 samples)
- Repeated station identifiers associated with different sampling events

These issues will be addressed systematically during the preprocessing stage in the following sections.

# Section 11 : Data Cleaning & Preprocessing

## Objective

The purpose of this section is to prepare the Indian MOREPOC dataset for remote sensing analysis while preserving as much valuable field information as possible.

Unlike conventional cleaning approaches that remove all incomplete observations, this workflow retains samples with missing geographic coordinates because these coordinates may be recovered later from original publications or supplementary metadata.

The preprocessing includes:

1. Remove observations with missing target variable (conc_poc)
2. Preserve observations with missing coordinates
3. Create a coordinate availability flag
4. Standardize date format
5. Create unique sample identifiers
6. Save the cleaned dataset

This approach maximizes data retention while ensuring transparency throughout the preprocessing workflow.

```python
# ============================================================
# Section 11.1 : Create Working Copy
# ============================================================

clean_df = morepoc.copy()

print("="*70)
print("Working Copy Created")
print("="*70)

print(clean_df.shape)
```

```python
# ============================================================
# Section 11.2 : Remove Missing POC
# ============================================================

before = len(clean_df)

clean_df = clean_df.dropna(subset=["conc_poc"])

after = len(clean_df)

print("="*70)
print("Missing Target Variable")
print("="*70)

print("Before :", before)
print("After  :", after)
print("Removed:", before-after)
```

```python
# ============================================================
# Section 11.3 : Coordinate Status
# ============================================================

clean_df["coordinate_status"] = np.where(

    clean_df["lat"].isna() |
    clean_df["lon"].isna(),

    "Missing",

    "Available"

)

print("="*70)
print("Coordinate Availability")
print("="*70)

print(clean_df["coordinate_status"].value_counts())
```

```python
# ============================================================
# Section 11.4 : Sample ID
# ============================================================

clean_df = clean_df.reset_index(drop=True)

clean_df["sample_id"] = [

    f"IND_{i+1:04d}"

    for i in range(len(clean_df))

]

display(

clean_df[
    [
        "sample_id",
        "riv_id",
        "sample_date",
        "coordinate_status"
    ]
].head()

)
```

```python
# ============================================================
# Section 11.5 : Date
# ============================================================

clean_df["sample_date"] = pd.to_datetime(
    clean_df["sample_date"]
)

print(clean_df["sample_date"].dtype)
```

```python
# ============================================================
# Section 11.6 : Final Dataset Summary
# ============================================================

print("="*70)
print("Clean Dataset Summary")
print("="*70)

print("Rows :", len(clean_df))
print("Columns :", len(clean_df.columns))

print("\nCoordinate Status")

print(clean_df["coordinate_status"].value_counts())

display(clean_df.head())
```

```python
# ============================================================
# Section 11.7 : Save
# ============================================================

clean_df.to_csv(

    "Indian_MOREPOC_Prepared.csv",

    index=False

)

print("="*70)
print("Dataset Saved Successfully")
print("="*70)

print("Indian_MOREPOC_Prepared.csv")
```

```python
import ee
import geemap

# Initialize Earth Engine
ee.Initialize(project="oc-flux")

# Create a map
Map = geemap.Map()

# Center on India
Map.setCenter(78.9629, 22.5937, 5)

Map
```

```python
landsat = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterDate("2023-01-01", "2023-12-31")
    .filterBounds(ee.Geometry.Point([77.2, 28.6]))  # Near Delhi
    .first()
)

vis = {
    "bands": ["SR_B4", "SR_B3", "SR_B2"],
    "min": 7000,
    "max": 18000,
}

Map.addLayer(landsat, vis, "Landsat 8")
Map
```

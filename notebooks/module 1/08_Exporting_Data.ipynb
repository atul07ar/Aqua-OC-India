```python
Export it:
```

```python
task = ee.batch.Export.image.toDrive(
    image=median,
    description="Median_Landsat_2023",
    folder="GEE_Exports",
    fileNamePrefix="median_2023",
    region=ee.Geometry.Point([77.2090,28.6139]).buffer(10000).bounds(),
    scale=30,
    crs="EPSG:4326",
    maxPixels=1e13
)

task.start()

print("Export Started!")
```

```python
# =====================================================
# Lesson 8 - Exporting Data
# =====================================================

import ee
import geemap

# Initialize Earth Engine
ee.Initialize(project="oc-flux")

# Study Area
point = ee.Geometry.Point([77.2090, 28.6139])

# Load Landsat Collection
collection = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
      .filterDate("2023-01-01", "2023-12-31")
      .filterBounds(point)
)

# Create Median Composite
median = collection.median()
```

```python
Creating a Median Composite
```

```python
import ee
import geemap

ee.Initialize(project="oc-flux")
```

```python
Load images.
```

```python
collection = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
      .filterDate("2023-01-01","2023-12-31")
      .filterBounds(
          ee.Geometry.Point([77.2090,28.6139])
      )
)
```

```python
Create composite.
```

```python
median = collection.median()
```

```python
Display it.
```

```python
Map = geemap.Map()

Map.centerObject(
    ee.Geometry.Point([77.2090,28.6139],
    ),
    9
)

Map.addLayer(
    median,
    {
        "bands":["SR_B4","SR_B3","SR_B2"],
        "min":7000,
        "max":18000
    },
    "Median Composite"
)

Map
```

```python
Mean vs Median

Create both.
```

```python
Display together
```

```python
mean = collection.mean()

median = collection.median()
```

```python
Map = geemap.Map()

Map.centerObject(
    ee.Geometry.Point([77.2090,28.6139]),
    9
)

Map.addLayer(
    mean,
    {
        "bands":["SR_B4","SR_B3","SR_B2"],
        "min":7000,
        "max":18000
    },
    "Mean"
)

Map.addLayer(
    median,
    {
        "bands":["SR_B4","SR_B3","SR_B2"],
        "min":7000,
        "max":18000
    },
    "Median"
)

Map
```

```python
def mask_clouds(image):

    qa = image.select("QA_PIXEL")

    cloud = qa.bitwiseAnd(1 << 3).eq(0)

    shadow = qa.bitwiseAnd(1 << 4).eq(0)

    mask = cloud.And(shadow)

    return image.updateMask(mask)
```

```python
clean_collection = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
      .filterDate("2023-01-01","2023-12-31")
      .filterBounds(ee.Geometry.Point([77.2090,28.6139]))
      .map(mask_clouds)
)
```

```python
Map = geemap.Map()

image = clean_collection.first()

Map.centerObject(
    ee.Geometry.Point([77.2090,28.6139]),
    9
)

Map.addLayer(
    image,
    {
        "bands":["SR_B4","SR_B3","SR_B2"],
        "min":7000,
        "max":18000
    },
    "Cloud Masked"
)

Map
```

```python
import ee
import geemap

ee.Initialize(project="oc-flux")

def mask_clouds(image):
    qa = image.select("QA_PIXEL")
    cloud = qa.bitwiseAnd(1 << 3).eq(0)
    shadow = qa.bitwiseAnd(1 << 4).eq(0)
    mask = cloud.And(shadow)
    return image.updateMask(mask)

collection = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
      .filterDate("2023-01-01","2023-12-31")
      .filterBounds(ee.Geometry.Point([77.2090,28.6139]))
      .map(mask_clouds)
)

image = collection.first()

Map = geemap.Map()
Map.centerObject(ee.Geometry.Point([77.2090,28.6139]),9)

Map.addLayer(
    image,
    {
        "bands":["SR_B4","SR_B3","SR_B2"],
        "min":7000,
        "max":18000
    },
    "Cloud Masked"
)

Map
```

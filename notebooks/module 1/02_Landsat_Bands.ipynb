```python
import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()

landsat = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterDate("2023-01-01", "2023-12-31")
    .filterBounds(ee.Geometry.Point([77.2, 28.6]))
    .first()
)

# True color
Map.addLayer(
    landsat,
    {"bands": ["SR_B4", "SR_B3", "SR_B2"], "min": 7000, "max": 18000},
    "True Color",
)

# False color
Map.addLayer(
    landsat,
    {"bands": ["SR_B5", "SR_B4", "SR_B3"], "min": 7000, "max": 18000},
    "False Color",
)

Map.setCenter(77.2, 28.6, 9)

Map
```

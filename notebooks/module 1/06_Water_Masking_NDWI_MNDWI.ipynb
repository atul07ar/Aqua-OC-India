```python
First load an image.
```

```python
import ee
import geemap

ee.Initialize(project="oc-flux")

image = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
      .filterDate("2023-01-01","2023-12-31")
      .filterBounds(
          ee.Geometry.Point([77.2090,28.6139])
      )
      .first()
)
```

```python
Now calculate NDWI.
```

```python
ndwi = image.normalizedDifference(
    ["SR_B3","SR_B5"]
).rename("NDWI")
```

```python
Display NDWI
```

```python
Map = geemap.Map()

Map.centerObject(
    ee.Geometry.Point([77.2090,28.6139]),
    9
)

Map.addLayer(
    ndwi,
    {
        "min":-1,
        "max":1,
        "palette":[
            "brown",
            "yellow",
            "white",
            "cyan",
            "blue"
        ]
    },
    "NDWI"
)

Map
```

```python
Dark Blue

River

NDWI = 0.7

Light Blue

Lake

NDWI = 0.4

Brown

Forest

NDWI = -0.3

Yellow

City

NDWI = -0.1

Positive values usually indicate water.

Negative values usually indicate land.
```

```python
water = ndwi.gt(0)
```

```python
Map = geemap.Map()

Map.centerObject(
    ee.Geometry.Point([77.2090,28.6139]),
    9
)

Map.addLayer(
    water,
    {
        "palette":["white","blue"]
    },
    "Water Mask"
)

Map
```

```python
Blue pixels = Water.

White pixels = Land.
```

```python
water_image = image.updateMask(water)
```

```python
Now only water remains.
```

```python
Map = geemap.Map()

Map.centerObject(
    ee.Geometry.Point([77.2090,28.6139]),
    9
)

Map.addLayer(
    water_image,
    {
        "bands":["SR_B4","SR_B3","SR_B2"],
        "min":7000,
        "max":18000
    },
    "Water Only"
)

Map
```

```python
You should now see mainly rivers and other water bodies.
```

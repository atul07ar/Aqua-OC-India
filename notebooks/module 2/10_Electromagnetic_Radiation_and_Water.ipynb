```python
import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()

point = ee.Geometry.Point([77.2090, 28.6139])

image = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
      .filterBounds(point)
      .filterDate("2023-01-01", "2023-12-31")
      .median()
)

# True Color
Map.addLayer(
    image,
    {
        "bands": ["SR_B4", "SR_B3", "SR_B2"],
        "min": 7000,
        "max": 18000
    },
    "True Color"
)

# Near Infrared Display
Map.addLayer(
    image,
    {
        "bands": ["SR_B5"],
        "min": 7000,
        "max": 18000
    },
    "NIR"
)

Map.centerObject(point, 10)

Map
```

```python
What to Observe
In the True Color layer, identify rivers or lakes.
Turn on the NIR layer.
Notice that water becomes much darker than surrounding vegetation.

This is direct evidence of water's strong absorption of NIR.
```

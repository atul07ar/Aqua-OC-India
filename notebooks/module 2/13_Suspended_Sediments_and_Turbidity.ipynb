```python
# ==========================================
# Lesson 13 - Suspended Sediments and Turbidity
# ==========================================

import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()

# Study area (Yamuna River near Delhi)
point = ee.Geometry.Point([77.2090, 28.6139])

collection = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(point)
    .filterDate("2023-01-01", "2023-12-31")
)

image = collection.median()

Map.centerObject(point, 10)

Map
```

```python
Practical Demonstration

We'll compare a natural-color image with a red-band image.
Observation

Use the layer control to switch between:

True Color
Red Band

Look for:

Rivers
Reservoirs
Bright or dark water
Areas that may contain suspended sediment
```

```python
Map = geemap.Map()

Map.centerObject(point, 10)

# True Color
Map.addLayer(
    image,
    {
        "bands": ["SR_B4", "SR_B3", "SR_B2"],
        "min": 7000,
        "max": 18000
    },
    "True Color"
)

# Red Band
Map.addLayer(
    image,
    {
        "bands": ["SR_B4"],
        "min": 7000,
        "max": 18000
    },
    "Red Band"
)

Map
```

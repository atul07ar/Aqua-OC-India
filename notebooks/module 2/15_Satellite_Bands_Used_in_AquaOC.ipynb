```python
# ==========================================
# Lesson 15 - Satellite Bands Used in Aqua-OC
# STANDARD SETUP RUN THIS FIRST
# ==========================================

import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()

# Study Area (Yamuna River)
point = ee.Geometry.Point([77.2090, 28.6139])

collection = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(point)
    .filterDate("2023-01-01", "2023-12-31")
)

image = collection.median()

Map.centerObject(point, 10)

Map
```

```python
Visualize Individual Bands
```

```python
Blue
```

```python
Map = geemap.Map()

Map.centerObject(point,10)

Map.addLayer(
    image,
    {
        "bands":["SR_B2"],
        "min":7000,
        "max":18000
    },
    "Blue"
)

Map
```

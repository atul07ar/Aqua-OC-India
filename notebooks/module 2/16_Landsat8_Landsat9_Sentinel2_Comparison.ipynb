```python
# ==========================================
# Lesson 16 - Landsat 8 vs Landsat 9 vs Sentinel-2
# ==========================================

import ee
import geemap

# Initialize Earth Engine
ee.Initialize(project="oc-flux")

# Create Map
Map = geemap.Map()

# Study Area (Yamuna River near Delhi)
point = ee.Geometry.Point([77.2090, 28.6139])

# Landsat 8 Collection
landsat8 = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(point)
    .filterDate("2023-01-01", "2023-12-31")
)

# Landsat 9 Collection
landsat9 = (
    ee.ImageCollection("LANDSAT/LC09/C02/T1_L2")
    .filterBounds(point)
    .filterDate("2023-01-01", "2023-12-31")
)

# Sentinel-2 Collection
sentinel = (
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(point)
    .filterDate("2023-01-01", "2023-12-31")
)

# Create Median Composites
landsat8_image = landsat8.median()
landsat9_image = landsat9.median()
sentinel_image = sentinel.median()

# Center Map
Map.centerObject(point, 10)

Map
```

```python
Map = geemap.Map()

landsat = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(point)
    .filterDate("2023-01-01","2023-12-31")
    .median()
)

Map.centerObject(point,10)

Map.addLayer(
    landsat,
    {
        "bands":["SR_B4","SR_B3","SR_B2"],
        "min":7000,
        "max":18000
    },
    "Landsat 8"
)

Map
```

```python
Sentinel-2
```

```python
Map = geemap.Map()

sentinel = (
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(point)
    .filterDate("2023-01-01","2023-12-31")
    .median()
)

Map.centerObject(point,10)

Map.addLayer(
    sentinel,
    {
        "bands":["B4","B3","B2"],
        "min":0,
        "max":3000
    },
    "Sentinel-2"
)

Map
```

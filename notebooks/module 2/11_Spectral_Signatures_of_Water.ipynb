```python
# ==========================================
# Lesson 11 - Spectral Signatures of Water
# ==========================================

import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()

# Study area (Delhi for demonstration)
point = ee.Geometry.Point([77.2090, 28.6139])

collection = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(point)
    .filterDate("2023-01-01", "2023-12-31")
)

image = collection.median()

Map.centerObject(point, 10)

Map
```

```python
Visual Comparison

Run this.
```

```python
Map = geemap.Map()

Map.centerObject(point, 10)

# True Color
Map.addLayer(
    image,
    {
        "bands": ["SR_B4", "SR_B3", "SR_B2"],
        "min": 7000,
        "max": 18000
    },
    "True Color"
)

# False Color
Map.addLayer(
    image,
    {
        "bands": ["SR_B5", "SR_B4", "SR_B3"],
        "min": 7000,
        "max": 18000
    },
    "False Color"
)

Map
```

```python
Observe:

True Color

Looks like a normal photograph.

False Color

Healthy vegetation appears bright red.
Water appears very dark.
Urban areas often appear cyan or gray.
```

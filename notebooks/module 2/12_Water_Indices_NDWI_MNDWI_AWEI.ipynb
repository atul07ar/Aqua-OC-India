```python
# ==========================================
# Lesson 12 - Water Indices
# ==========================================

import ee
import geemap

ee.Initialize(project="oc-flux")

Map = geemap.Map()

# Study Area (Delhi/Yamuna River)
point = ee.Geometry.Point([77.2090, 28.6139])

collection = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(point)
    .filterDate("2023-01-01", "2023-12-31")
)

image = collection.median()

Map.centerObject(point, 10)

Map
```

```python
NDWI in Earth Engine

Run:
Observe
Blue = Water
Brown = Land
White = Transition
```

```python
ndwi = image.normalizedDifference(["SR_B3", "SR_B5"])

Map = geemap.Map()

Map.centerObject(point, 10)

Map.addLayer(
    ndwi,
    {
        "min": -1,
        "max": 1,
        "palette": ["brown", "white", "blue"]
    },
    "NDWI"
)

Map
```

```python
Calculate MNDWI
```

```python
mndwi = image.normalizedDifference(["SR_B3", "SR_B6"])

Map = geemap.Map()

Map.centerObject(point,10)

Map.addLayer(
    mndwi,
    {
        "min": -1,
        "max": 1,
        "palette": ["brown","white","blue"]
    },
    "MNDWI"
)

Map
```

```python
Calculate AWEI
```

```python
awei = image.expression(
    '4*(GREEN-SWIR1) - (0.25*NIR + 2.75*SWIR2)',
    {
        'GREEN': image.select('SR_B3'),
        'NIR': image.select('SR_B5'),
        'SWIR1': image.select('SR_B6'),
        'SWIR2': image.select('SR_B7')
    }
)

Map = geemap.Map()

Map.centerObject(point,10)

Map.addLayer(
    awei,
    {
        "min": -5000,
        "max": 5000,
        "palette": ["brown","white","blue"]
    },
    "AWEI"
)

Map
```

```python
Compare All Three
```

```python
Map = geemap.Map()

Map.centerObject(point,10)

Map.addLayer(ndwi,{"min":-1,"max":1,"palette":["brown","white","blue"]},"NDWI")

Map.addLayer(mndwi,{"min":-1,"max":1,"palette":["brown","white","blue"]},"MNDWI")

Map.addLayer(awei,{"min":-5000,"max":5000,"palette":["brown","white","blue"]},"AWEI")

Map
```
